```python
import pandas as pd

data = {
    'student_name':     ['EMU', 'EMU', 'demon', 'lemon', 'emon'],
    'hours_studied':    [5, 8, 8, 10, 6],
    'previous_score':   [55, 75, 75, 85, 95],
    'attendance':       [4, 5, 6, 6, 8],
    'sleep_hours':      [6, 7, 8, 8, 10],
    'extracurricular':  ['yes', 'no', 'yes', 'no', 'yes'],
    'parent_education': ['bachelors', 'masters', '+2', 'PhD', 'masters']
}

df = pd.DataFrame(data)
print("Student Performance DataFrame:")
print(df)

print("\nData Info:")
print(df.info())

print("\nDataFrame shape:", df.shape[0], "rows", df.shape[1], "columns")
print("Column names:", df.columns.tolist())
print("Data:")
print(df)
```

```python
print("first 3 row:")
print(df.head(3))

print("last 2 row:")
print(df.tail(2))

print("statistical summary:")
print(df.describe())

print("missing value:")
print(df.isnull().sum())

print("check for NaN values:")
print(df.isna().sum())

print("duplicate rows:")
print(df.duplicated().sum())

print("student nAme only:")
print(df('student_name'))




```
